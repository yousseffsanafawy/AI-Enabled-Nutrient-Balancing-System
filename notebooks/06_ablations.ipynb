## 1. Executive Summary & Ablation Objectives

To defend our proposed **Multi-Task Dilated Temporal Convolutional LSTM Network (MT-TCN-LSTM)** for peer review, we conduct systematic ablation experiments across three distinct axes:
1. **Architectural Components:**
   - Does the dilated causal Conv1D front-end contribute significantly over a pure recurrent LSTM?
   - How does GRU compare against LSTM for multi-task sequence aggregation?
   - Does joint training with the auxiliary stress head ($\lambda = 0.2$) regularize or hinder continuous regression?
2. **Lookback Window Size Horizon ($W \in \{5, 10, 15, 30, 60\}$ steps):**
   - What is the empirical trade-off between physical lag capture and high-frequency noise sensitivity?
3. **Feature Selection & Normalization:**
   - How does forecasting change when restricting the model to chemical sensors only (pH, TDS)?
   - Does `MinMaxScaler` outperform `StandardScaler` (Z-score) on bounded IoT sensor ranges?

## 4. Key Scientific Insights for the Research Paper

1. **The Role of Causal Dilations:**
   Causal dilated 1D convolutions expand the temporal receptive field to capture nutrient mixing trends without contracting the sequence dimension, outperforming pure recurrent baselines.
2. **Multi-Task Synergistic Effect:**
   Adding the auxiliary stress classification head ($\lambda = 0.2$) acts as domain-aware regularization, improving boundary sensitivity without degrading continuous tracking.
3. **Optimal Horizon ($W=15$ steps / 150 seconds):**
   A 15-step lookback provides the ideal balance between short-term noise filtering and physical hydraulic response time.
