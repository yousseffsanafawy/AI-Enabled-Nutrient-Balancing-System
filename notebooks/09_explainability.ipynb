## 2. Load Structured Explainability Experiment Log (`exp_009_explainability.json`)

```python
exp_path = os.path.join(ROOT_DIR, 'experiments', 'exp_009_explainability.json')
with open(exp_path, 'r') as f:
    exp_data = json.load(f)

print(f"Experiment ID: {exp_data['experiment_id']}")
print(f"Model: {exp_data['model_architecture']}")
print(f"Evaluation Samples: {exp_data['sample_evaluation_size']}")

# Display Permutation Importance Table
perm_info = exp_data['permutation_feature_importance']['feature_importance']
rows = []
for feat, d in perm_info.items():
    rows.append({
        'Sensor Modality': feat,
        'Relative Importance Share (%)': f"{d['normalized_importance']*100:.2f}%",
        'Mean MSE Increase': f"{d['mean_mse_increase']:.6f}",
        'Std MSE Increase': f"{d['std_mse_increase']:.6f}",
        'Stress ROC-AUC Drop': f"{d['stress_auc_drop']:.4f}",
    })
df_perm = pd.DataFrame(rows)
df_perm
```
